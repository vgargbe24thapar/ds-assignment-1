```python
import numpy as np
import matplotlib.pyplot as plt
from skimage.util import random_noise

M,N = 500, 500
U = np.random.rand(M, N)
#1.uniform
a_unif, b_unif = 10, 50
uniform_noise = a_unif + (b_unif - a_unif) * U
#2.gaussian
mu, sigma = 0, 10
U1 = np.random.rand(M, N)
U2 = np.random.rand(M,N)
gaussian_noise_custom = mu + sigma * (np.sqrt(-2 * np.log(U1)) * np.cos(2 * np.pi * U2))

blank_image = np.zeros((M,N))
gaussian_noise_inbuilt = random_noise(blank_image, mode='gaussian', mean=mu, var=(sigma/255)**2)
gaussian_noise_inbuilt = gaussian_noise_inbuilt * 255

# 3. Erlang (Gamma) Noise
a_erlang, b_erlang = 2, 5
erlang_prod = np.ones((M, N))

for _ in range(b_erlang):
    erlang_prod *= np.random.rand(M, N)

erlang_noise = (-1 / a_erlang) * np.log(erlang_prod)

# 4. Exponential Noise
a_exp = 1
exponential_noise = (-1 / a_exp) * np.log(1 - U)

# 5. Rayleigh Noise
a_ray, b_ray = 0, 10
rayleigh_noise = a_ray + np.sqrt(
    -b_ray * np.log(1 - U)
)

  # ---- Plotting Histograms ----
fig, axes = plt.subplots(2, 3, figsize=(15, 10))
fig.suptitle('Histograms of Generated Noise Distributions', fontsize=16)

# Helper function to plot histograms
def plot_hist(ax, data, title):
    ax.hist(data.flatten(), bins=50, density=True, alpha=0.7, color='blue')
    ax.set_title(title)
    ax.grid(True, alpha=0.3)

plot_hist(axes[0, 0], uniform_noise, '1. Uniform Noise')
plot_hist(axes[0, 1], gaussian_noise_custom, '2. Gaussian Noise (Custom)')
plot_hist(axes[0, 2], erlang_noise, '3. Erlang Noise')
plot_hist(axes[1, 0], exponential_noise, '4. Exponential Noise')
plot_hist(axes[1, 1], rayleigh_noise, '5. Rayleigh Noise')
plot_hist(axes[1, 2], gaussian_noise_inbuilt, 'Gaussian Noise (Inbuilt sk-image)')

plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()

```
