```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

img = cv2.imread("grayscale1.jpg", cv2.IMREAD_GRAYSCALE)


threshold = np.mean(img)
binary = np.where(img >= threshold, 255, 0).astype(np.uint8)

print("Mean Threshold =", threshold)

cv2_imshow(img)
cv2_imshow(binary)

```
